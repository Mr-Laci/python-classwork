```python
import pandas as pd
#Wednesday
#Question 1. Load and inspect
from google.colab import drive
drive.mount('/content/drive')
df = pd.read_csv("/content/drive/MyDrive/suya_orders.csv")
print(f'\n This Record and Fields{df.shape}\n')
print(df.head())
print(df.info())

#Question 2. Print the Item column
print(f"\n{df["Item"]}")

#Question 3. Orders from "Kaduna Central"
df = df[df["Branch"] == "Kaduna Central"]
print(f"\n{df}\n")
print(f"\nNumber of orders from Kaduna Central:\n {len(df)}\n")

#Question 4. Save first 100 rows
df.head(100).to_csv("\norders_100.csv\n", index=False)
print("\nSaved first 100 rows to orders_100.csv")
```

```python
#Week 5
#Monday: Question 1
df = df.copy()
print("Missing values in each column:")
print(df.isnull().sum())

# 2. Fill missing values
df["Customer_Age"] = df["Customer_Age"].fillna(df["Customer_Age"].median())
df["Payment_Method"] = df["Payment_Method"].fillna("Unknown")
print("\nMissing values after filling:")
print(df.isnull().sum())

# 3. Count and remove duplicate rows
print(f"\nNumber of duplicate rows: {df.duplicated().sum()}")
df = df.drop_duplicates()
print(f"after cleaning duplicates: {df.shape}")

# 4. Convert Date column to datetime
df["Date"] = pd.to_datetime(df["Date"])
print(f"\nDate column type: {df["Date"].dtype}")
print(df["Date"].head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

 #Wednesday: Question 1
top_10 = df.nlargest(10, "Total_NGN")
print("10 biggest orders:")
print(top_10[["Order_ID", "Item", "Quantity", "Total_NGN", "Branch"]])

# Look for strange ones (very high Quantity or unrealistic totals)
print("\nPossible strange orders (Quantity > 20):")
print(top_10[top_10["Quantity"] > 20][["Order_ID", "Item", "Quantity", "Total_NGN"]])

# 2. Boxplot of Total_NGN
plt.figure(figsize=(8, 5))
sns.boxplot(x=df["Total_NGN"])
plt.title("Boxplot of Total_NGN")
plt.show()

# 3. Create a new Month column from Date
df["Month"] = df["Date"].dt.month
print("\nMonth column sample:")
print(df[["Date", "Month"]].head())

# 4. Change Is_Delivery from Yes/No to 1/0
df["Is_Delivery"] = df["Is_Delivery"].map({"Yes": 1, "No": 0})
print("\nIs_Delivery after conversion:")
print(df["Is_Delivery"].value_counts())
```

```python
#Week 6
#Monday: Question 1
print(df["Total_NGN"].describe())

# 2. Average Total_NGN for each Branch
print(df.groupby("Branch")["Total_NGN"].mean())

# 3. Correlation between Quantity and Total_NGN
print(df["Quantity"].corr(df["Total_NGN"]))

# 4. Mean and Median of Total_NGN
mean_val = df["Total_NGN"].mean()
median_val = df["Total_NGN"].median()
print("Mean:", mean_val)
print("Median:", median_val)
print("Mean is bigger" if mean_val > median_val else "Median is bigger")

# 5. Count of orders for each Item
print(df["Item"].value_counts())

# 6. Total sales for each Payment_Method
print(df.groupby("Payment_Method")["Total_NGN"].sum())

# 7. Average Rating for each Branch
print(df.groupby("Branch")["Rating"].mean())

# 8. Correlation between Customer_Age and Total_NGN
print(df["Customer_Age"].corr(df["Total_NGN"]))

# 9. Correlation between Quantity and Rating
print(df["Quantity"].corr(df["Rating"]))

# 10. Percentage of delivery vs non-delivery orders
print(df["Is_Delivery"].value_counts(normalize=True) * 100)
```

```python
#Wednesday
import matplotlib.pyplot as plt
import seaborn as sns

df["Date"] = pd.to_datetime(df["Date"])
df["Month"] = df["Date"].dt.month

#Question 1. Bar chart – total sales per Branch
branch_sales = df.groupby("Branch")["Total_NGN"].sum().sort_values(ascending=False)
plt.figure(figsize=(10,5))
branch_sales.plot(kind="bar", color="orange")
plt.title("Total Sales by Branch")
plt.ylabel("Total Sales (₦)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

#Question 2. Seaborn boxplot – Total_NGN by Item
plt.figure(figsize=(10,5))
sns.boxplot(data=df, x="Item", y="Total_NGN")
plt.title("Total_NGN Distribution by Item")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

#Question 3. Line chart – total sales per Month
monthly_sales = df.groupby("Month")["Total_NGN"].sum()
plt.figure(figsize=(8,5))
monthly_sales.plot(kind="line", marker="o")
plt.title("Total Sales by Month")
plt.xlabel("Month")
plt.ylabel("Total Sales (₦)")
plt.grid(True)
plt.show()

#Question 4. Bar chart – number of orders per Item
item_counts = df["Item"].value_counts()
plt.figure(figsize=(8,5))
item_counts.plot(kind="bar", color="skyblue")
plt.title("Number of Orders by Item")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

#Question 5. Histogram of Customer_Age
plt.figure(figsize=(8,5))
plt.hist(df["Customer_Age"].dropna(), bins=20, color="green", edgecolor="black")
plt.title("Distribution of Customer Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()

#Question 6. Bar chart – average Rating per Branch
avg_rating = df.groupby("Branch")["Rating"].mean().sort_values(ascending=False)
plt.figure(figsize=(10,5))
avg_rating.plot(kind="bar", color="purple")
plt.title("Average Rating by Branch")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

#Question 7. Seaborn countplot of Payment_Method
plt.figure(figsize=(8,5))
sns.countplot(data=df, x="Payment_Method", order=df["Payment_Method"].value_counts().index)
plt.title("Orders by Payment Method")
plt.show()

#Question 8. Scatter plot – Quantity vs Total_NGN
plt.figure(figsize=(8,5))
plt.scatter(df["Quantity"], df["Total_NGN"], alpha=0.5)
plt.title("Quantity vs Total_NGN")
plt.xlabel("Quantity")
plt.ylabel("Total_NGN")
plt.show()

#Question 9. Correlation heatmap
corr_cols = ["Quantity", "Total_NGN", "Customer_Age", "Rating"]
plt.figure(figsize=(7,5))
sns.heatmap(df[corr_cols].corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()

#Question 10. Line chart – number of orders per Month
monthly_orders = df.groupby("Month").size()
plt.figure(figsize=(8,5))
monthly_orders.plot(kind="line", marker="o", color="red")
plt.title("Number of Orders by Month")
plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.grid(True)
plt.show()
```
